7. If Statement

8. If-Else

```r
marks <- 75

if(marks >= 50){
  print("Pass")
} else {
  print("Fail")
}
```

9. For Loop

```r
for(i in 1:5){
  print(i)
}
```

10. While Loop

```r
i <- 1

while(i <= 5){
  print(i)
  i <- i + 1
}
```

14. Data Frame

```r
students <- data.frame(
  Name = c("Arun", "Bala", "Charan"),
  Marks = c(90, 85, 95)
)

print(students)
```

15.Bar Chart

```r
marks <- c(80, 90, 70, 95)

barplot(
  marks,
  names.arg = c("Math", "Science", "English", "Computer"),
  col = c("red", "blue", "green", "orange"),
  main = "Student Marks",
  xlab = "Subjects",
  ylab = "Marks"
)
```
