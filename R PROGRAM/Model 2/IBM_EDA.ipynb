# IBM Employee Dataset - Exploratory Data Analysis (EDA)

## Overview

The IBM Employee Dataset contains information about employees working in an organization. It includes employee details such as age, education, department, monthly income, job satisfaction, work-life balance, years at the company, and attrition status.

This dataset is commonly used for Exploratory Data Analysis (EDA) R Language to understand employee characteristics and identify factors that may influence employee attrition.


```r
# Load Dataset
data <- read.csv("IBM.csv")
```

## Dataset Features

- Age
- Attrition
- Department
- DistanceFromHome
- Education
- EducationField
- EnvironmentSatisfaction
- JobSatisfaction
- MaritalStatus
- MonthlyIncome
- NumCompaniesWorked
- WorkLifeBalance
- YearsAtCompany

## Task 5: Outlier Detection

#### This task uses boxplots to identify outliers in important numerical attributes such as Age, Monthly Income, and Years at Company.

```r
# Figure Size
options(repr.plot.width = 18, repr.plot.height = 9)

# Monthly Income - Box Plot
boxplot(data$MonthlyIncome,
        main = "Box Plot of Monthly Income",
        xlab = "Employees",
        ylab = "Monthly Income",
        col = "Yellow",
        border = "darkblue",
        notch = TRUE,
        outline = TRUE,
        cex.main = 1.2,
        cex.lab = 1,
        cex.axis = 0.9)
```

## Task 6: Bivariate Analysis

#### This task examines the relationship between employee attrition and categorical variables such as Gender, Department, Job Role, and Marital Status.

```r
# Load Libraries
library(ggplot2)
library(dplyr)
```

```r

options(repr.plot.width = 38, repr.plot.height = 30)

# Attrition vs Gender
barplot(table(data$Age,data$Department),
        beside=TRUE,
        col = c("skyblue", "tomato"),
        main = "Age vs Department",
        xlab = "Age",
        ylab = "Department",
        legend.text = TRUE)
```

```r
# -----------------------------
# Attrition vs Department
# -----------------------------
options(repr.plot.width = 38, repr.plot.height = 30)

barplot(table(data$Attrition, data$Department),
        beside = TRUE,
        col = c("skyblue", "tomato", "lightgreen"),
        main = "Attrition vs Department",
        xlab = "Attrition",
        ylab = "Number of Employees",
        legend.text = TRUE)
```

```r
# -----------------------------
# Attrition vs Job Role
# -----------------------------
options(repr.plot.width = 38, repr.plot.height = 30)
barplot(table(data$Attrition, data$JobSatisfaction),
        beside = TRUE,
        col = rainbow(length(unique(data$JobSatisfaction))),
        main = "Attrition vs JobSatisfaction",
        xlab = "Attrition",
        ylab = "JobSatisfactions",
        legend.text = TRUE,
        las = 2,
        cex.names = 0.7)
```

```r
# -----------------------------
# Attrition vs Marital Status
# -----------------------------
options(repr.plot.width = 38, repr.plot.height = 30)
barplot(table(data$Attrition, data$MaritalStatus),
        beside = TRUE,
        col = c("lightgreen", "lightblue", "salmon"),
        main = "Attrition vs Marital Status",
        xlab = "Attrition",
        ylab = "Number of Employees",
        legend.text = TRUE)
```

## Task 7: Numerical Relationships

#### This task uses scatter plots to analyze the relationship between numerical variables such as Age, Monthly Income, and Total Working Years.

```r
# Figure Size (Jupyter Notebook)
options(repr.plot.width = 18, repr.plot.height = 16)

# Set Colors for Attrition
attrition_col <- ifelse(data$Attrition == "Yes", "red", "blue")

# Age vs Monthly Income
plot(data$Age,data$MonthlyIncome,
     pch = 19,
     col = attrition_col,
     main = "Age vs Monthly Income",
     xlab = "Age",
     ylab = "Monthly Income",
     cex = 0.8)

legend("topleft",
       legend = c("Yes", "No"),
       col = c("red", "blue"),
       pch = 19,
       title = "Attrition")
```

```r
# Figure Size (Jupyter Notebook)
options(repr.plot.width = 25, repr.plot.height = 19)

# Set Colors for Attrition
attrition_col <- ifelse(data$Attrition == "Yes", "Orange", "Green")

# Years At Company vs Monthly Income
plot(data$YearsAtCompany,data$MonthlyIncome,
     pch = 19,
     col = attrition_col,
     main = "Years at Company vs Monthly Income",
     xlab = "Years at Company",
     ylab = "Monthly Income",
     cex = 0.8)

legend("topleft",
       legend = c("Yes", "No"),
       col = c("Orange", "Green"),
       pch = 19,
       title = "Attrition")


```
